# P1 · 05 Manuscript tables and figures

Rebuilds every table and figure of the manuscript from the saved result files in `results/` (no GPU, about 2 minutes). The tables are written as Markdown snippets (`T_*.md`) and the figures as PNG/PDF/SVG. `make_tables.py` also recomputes the H1/H3 paired differences from the per-seed rows and writes `check_H_tests.csv`.

```python
import os, subprocess, sys
from pathlib import Path
try:
    from google.colab import drive; drive.mount('/content/drive')
    CODE_DIR = Path('/content/drive/MyDrive/P1_code_release')
except ImportError:
    CODE_DIR = Path('..').resolve()
OUT = Path('/content/P1_tables_figures') if Path('/content').exists() else Path('tables_figures')
OUT.mkdir(exist_ok=True)
env = dict(os.environ, P1_RESULTS=str(CODE_DIR / 'results'))
for s in ['make_tables_main.py', 'make_tables.py', 'make_figures.py', 'make_fig1_study_design.py']:
    r = subprocess.run([sys.executable, str(CODE_DIR / 'analysis' / s)], cwd=OUT, env=env, capture_output=True, text=True)
    print(s, 'exit', r.returncode, r.stderr[-500:])
# Architecture diagrams (Figs. 2-5) are written next to the script
r = subprocess.run([sys.executable, str(CODE_DIR / 'analysis' / 'make_diagrams.py')], env=env, capture_output=True, text=True)
print('make_diagrams.py exit', r.returncode)
print(sorted(p.name for p in OUT.iterdir()))
```
